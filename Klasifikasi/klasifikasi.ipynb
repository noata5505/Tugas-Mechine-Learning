​2. Student Depression Detection Dataset
https://www.kaggle.com/datasets/mkhoirulmustaqim/student-depression-detection-dataset?hl=id-ID
​Deskripsi: Dataset tabular yang populer digunakan untuk melakukan klasifikasi kesehatan mental (deteksi tingkat depresi pada pelajar berdasarkan berbagai faktor pendukung).

​8. Telecom Customer Churn Dataset
https://www.kaggle.com/datasets/alipakzad/telecom-customer-churn?hl=id-ID
​Deskripsi: Dataset klasik dunia telekomunikasi yang diperbarui dengan format modern untuk keperluan analisis prediktif, sangat direkomendasikan untuk klasifikasi pelanggan yang berpotensi berhenti berlangganan (churn) atau tetap loyal.
