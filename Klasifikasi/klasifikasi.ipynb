### 1. Inisialisasi & Membaca Dataset
Pertama, kita memuat dataset ke dalam Pandas DataFrame untuk mengevaluasi struktur kolomnya secara keseluruhan.

```python
import pandas as pd

# Load dataset
df = pd.read_csv('Enterprise_Cyber_Kill_Chain_Dataset.csv')

print(f"Dataset memiliki {df.shape[0]} baris dan {df.shape[1]} kolom.")
# Menampilkan semua nama kolom untuk melihat potensi variabel target
df.head()
```

### 2. Klasifikasi Biner (Binary Classification)
Dataset ini mendukung klasifikasi biner jika tujuannya adalah memprediksi hasil akhir (*outcome*) yang hanya memiliki dua kemungkinan, seperti apakah sebuah serangan berhasil atau tidak. Kolom **`Attack_Success`** sangat ideal sebagai target biner karena hanya berisi nilai `0` (Gagal) dan `1` (Berhasil).

```python
target_binary = 'Attack_Success'

print(f"--- Analisis Biner: Kolom '{target_binary}' ---")
print(f"Jumlah kelas unik: {df[target_binary].nunique()}")
print("\nDistribusi persentase kelas:")
print(df[target_binary].value_counts(normalize=True) * 100)
```
